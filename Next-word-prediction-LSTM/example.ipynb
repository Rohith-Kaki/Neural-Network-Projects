##### testing how the texts_to_sequnces work

```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer

texts = [
    'I love dogs',
    'I hate cats',
    'Dogs are loyal animals'
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

tokenizer.word_index
```

```python
sequences = tokenizer.texts_to_sequences(texts)[0]
sequences
texts = [
    'I love dogs',
    'I hate cats',
    'Dogs are loyal animals'
]
```

```python
l = []
for line in texts:
    sequnce = tokenizer.texts_to_sequences([line])[0]
    print(sequnce)
    print(len(sequnce))
    for i in range(0,len(sequnce)):
        l.append(sequnce[:i+1])
    break
```
